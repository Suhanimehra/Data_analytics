# Assignment 1
A temperature sensor ends conntinious readings between 20.0 degress Celsius and 25.0 degrees Celsius.
- Generate 1000 synthetic sensor readings
- compute the mean and median f your generated dataset
- generate a graphical visualization


```python
import numpy as np
import matplotlib.pyplot as plt

sensor_readings=np.random.uniform(
    low=20.0,
    high=25.0,
    size=1000
)

mean_sensor=np.mean(sensor_readings)
median_sensor=np.median(sensor_readings)

print(mean_sensor)
print(median_sensor)

plt.figure(figsize=(10,4))

plt.plot(
    sensor_readings,
    alpha=0.5,
    label='sensor readings',
    linewidth=2,
    zorder=1,
    color='red'
    
)

plt.axhline(
    mean_sensor,
    alpha=0.5,
    zorder=2,
    marker='o',
    markersize=10,
    label='mean',
    color='blue',
    linestyle='--',
    linewidth=2,
)

plt.axhline(
    median_sensor,
    alpha=0.5,
    zorder=2,
    marker='x',
    markersize=10,
    label='median',
    color='pink',
    linestyle='--',
    linewidth=2,
)

plt.title("SEnsor Readings")
plt.xlabel("Sensor Reading")
plt.ylabel("Temprature")
plt.legend()
plt.tight_layout()
plt.show()
```

# Assignment 2

E commerce / Web Analytics (A/B Testing Simulator)
- Task : Write a scriptt to simulate 1,000 incoming website visitors
- Assign each visitor a random float between 0.0 and 1.0. If the value is <0.5,route them to 'Design A'.
- Otherwise , route them to 'Design B'.
- print the percentage of visitors assigned to each design.

```python
import numpy as np 

visitor=np.random.uniform(
    low=0.0,
    high=1.0,
    size=1000
)

design_a=visitor[visitor<0.5]
design_b=visitor[visitor>=0.5]

# print(design_a)
# print(design_b)

percentage_a=(len(design_a)/len(visitor))*100
percentage_b=(len(design_b)/len(visitor))*100

print(percentage_a)
print(percentage_b)
```

# Assignment 4: E-Commerce Delivery Time /Estimator

Scenario: An e-commerce platform estimates order delivery times. On anverage, a delivery takes 30 minutes with a standard deviation of 5 minutes.

Tasks:
- 1. Simluate delivery times for 1000 orders.
- 2. Calculate what percentage of deliveries took longer than 40 minutes.
- 3. Plot the delivery times and draw a threshold at the 40 minute mark.

```python
import numpy as np

delivery_time=np.random.normal(loc=30.0,scale=5.0,size=1000)

longer_time=delivery_time[delivery_time>40]

percentage_delivery=(len(longer_time)/len(delivery_time))*100

print(percentage_delivery)

plt.figure(figsize=(10,4))

plt.plot(
    delivery_time,
    alpha=0.5,
    color='blue',
    zorder=1,
    linewidth=2
)

plt.axhline(
    y=40,
    linestyle='--',
    linewidth=3,
    alpha=0.5,
    zorder=4,
    label='Threshold',
    color='red'
    
)

plt.xlabel("Deliveries")
plt.ylabel("Time")
plt.legend()
plt.tight_layout()
plt.show()
```
