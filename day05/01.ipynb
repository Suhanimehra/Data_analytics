# Deterministic Domain and Grid Generators

### setup imports & plotting configuration

```python
import numpy as np 
import matplotlib.pyplot as plt

#apply clean visualization style
plt.style.use('seaborn-v0_8_whitegrid' if 'seaborn-v0_8_whitegrid' in plt.style.available else 'default')

#seaborn-v0_8_whitegrid: a clean white grid style inspired by seaborns's whitegrid theme.
plt.rcParams["figure.dpi"] = 100
#RC= runtime configuration 
```

## 1 Linear Spacinf('np.linespace')

```python
#retstep = retstep=True tells NumPy to also return the step size (the gap between consecutive numbers).
x_linear , step_size = np.linspace(start=0.0,stop=10.0,num=15,retstep=True)
print("Linear Space Array: ", x_linear)
print("Step Size: ", step_size)

plt.figure(figsize=(9,3)) #figsize=(width,height) in inches

plt.scatter(x_linear,np.zeros_like(x_linear),color='blue',s=75,zorder=3,label='Linear Space') #s=marker size


plt.vlines(x_linear, ymin=-0.1, ymax=0.1, color='blue', alpha=0.3, linestyle='--')
#alpha=transparency``

plt.title("1D unifrom grid linear space with vertical lines", fontsize=16)
plt.xlabel("X-axis", fontsize=14)
plt.yticks([]) #remove y-ticks
plt.ylim(-0.5,0.5)
plt.legend(loc='upper right', fontsize=12)
plt.tight_layout() #adjust subplots to fit into figure area.
plt.show()
```

# Logarithmic Spacing(np.logspace)

```python
# Generate 7 points from 10^-5(0.00001) to 10^1(10) using logarithmic space

learning_rates=np.logspace(start=5,stop=1,num=7,base=10)
print("Logarithimix Space Array: ", learning_rates)

# for i , lr in enumerate(learning_rates):
#     print(f"Learning Rate {i}: {lr:.6f}")

print("Learning Rates",[f"{lr:.6f}" for lr in learning_rates]) 

fig,(ax1,ax2)=plt.subplots(nrows=2,ncols=1,figsize=(9,4.5))

# top-plot : logarithmic visualization

ax1.scatter(learning_rates,np.zeros_like(learning_rates),color='red',s=60,zorder=3) #s=marker size
ax1.set_xscale('log') #set x-axis to logarithmic scale
ax1.set_title("Logarithmic Space Visualization (equal spacing in order of magnitude)", fontsize=16)
ax1.set_xlabel("Values (log scale)", fontsize=14)
ax1.set_yticks([]) #remove y-ticks

#bottom plot : linear visualization
ax2.scatter(learning_rates,np.zeros_like(learning_rates),color='red',s=60,zorder=3,label='Learning Rates') #s=marker size
ax2.set_title("Linear Space Visualization (actual spacing)", fontsize=16)
ax2.set_xlabel("Values (linear scale)", fontsize=14)
ax2.set_yticks([]) #remove y-ticks

plt.tight_layout() #adjust subplots to fit into figure area.
plt.show()
```

# Geometric Spacing


```python
geom_grid = np.geomspace(start=1.0,stop=1000.0 , num=6)

print("Geometric Space Array: ", geom_grid)
print(f"Common multiplicative ration of geometric space array: {geom_grid[1]/geom_grid[0]:.4f}")

plt.figure(figsize=(9,3.5)) #figsize=(width,height) in inches

plt.plot(range(len(geom_grid)),geom_grid, marker='o',color='green',markersize=8,linestyle='--',linewidth=2,label="np.geomspace(1 to 1000)") #s=marker size

plt.title("Geometric Space Visualization", fontsize=16)
plt.xlabel("Index", fontsize=14)    
plt.ylabel("Values", fontsize=14)
plt.yscale('log') #set y-axis to logarithmic scale

plt.grid(True,which='both',linestyle='--',linewidth=0.5,alpha=0.5) #grid lines for both major and minor ticks
plt.legend()
plt.tight_layout() #adjust subplots to fit into figure area.
plt.show()
```

# Step-Based Range(np.arange)

```python
time_steps=np.arange(start=0.0,stop=5.0,step=0.5)
print("Time Steps Array: ", time_steps)

print("Common difference of time steps array: ", time_steps[1]-time_steps[0])
print("Total number of time steps: ", len(time_steps))


#Graphical Visualization of time steps
plt.figure(figsize=(9,2.5)) #figsize=(width,height) in inches

#Stem plot Details:
# Stem plot: a discrete plot that displays data points as stems extending from a baseline along the x-axis. Each stem represents a data point, and the height of the stem corresponds to the value of that point. It is useful for visualizing discrete data or events over time.
#linefmt: "C4-" specifies the color and style of the stem lines (C4 is a color from the default color cycle, and '-' indicates a solid line).
#color cycle definations: C1-blue, C2-orange, C3-green, C4-red, C5-purple, C6-brown, C7-pink, C8-gray, C9-olive, C10-cyan

#markerfmt: "C4o" specifies the color and style of the markers at the top of each stem (C4 is red, and 'o' indicates circular markers).
#basefmt: "k-" specifies the color and style of the baseline (k is black, and '-' indicates a solid line).


plt.stem(time_steps,np.ones_like(time_steps),linefmt='C4--',markerfmt='C4o',basefmt='')

plt.title("Time Steps Visualization", fontsize=16)
plt.xlabel("Time Steps", fontsize=14)
plt.yticks([]) #remove y-ticks
plt.ylim(0,1.3)
plt.tight_layout() #adjust subplots to fit into figure area.
plt.show() 
```

# Coordinate Grid(np.meshgrid)



```python
x_domain=np.linspace(-2,2,20)
y_domain = np.linspace(-2,2,20)

#Construct 2D coordinate matrix
X,Y = np.meshgrid(x_domain,y_domain)

#Evaluate a 2D paraboloid surface Z=X^2 + Y^2
Z=X**2 + Y**2

print("Grid Coordinate output")
print(f"X Matrix",{X.shape})
print(f"Y Matrix",{Y.shape})
print(f"Z Matrix",{Z.shape})
print("sample X matrix slice(first 3 X 3 block):\n",X[:3,:3])
print("sample Y matrix slice(first 3 X 3 block):\n",Y[:3,:3])
print("sample Z matrix slice(first 3 X 3 block):\n",Z[:3,:3])


#Graphical Visualization of 2D paraboloid surface
fig , (ax1,ax2) = plt.subplots(nrows=1,ncols=2,figsize=(11,4.5))
# Subplot 1
ax1.scatter(X,Y,color='blue',marker='o',s=60,edgecolor='k') #s=marker size
ax1.set_title("2D Coordinate Grid mesh(np.meshgrid)", fontsize=11)
ax1.set_xlabel("X-axis", fontsize=11)
ax1.set_ylabel("Y-axis", fontsize=11)   
ax1.set_aspect('equal', ) #set equal aspect ratio for x and y axes

# Subplot 2
countour = ax2.contourf(X,Y,Z,cmap='viridis',levels=15) #filled contour plot

# ax2.contour(X,Y,Z,colors='k',levels=15,linewidths=0.5) #add contour lines to the filled contour plot

fig.colorbar(countour, ax=ax2, label='Z values = f(X,Y)') #add colorbar to the contour plot
ax2.set_title("2D Contour Plot of Paraboloid Surface", fontsize=11)
ax2.set_xlabel("X-axis", fontsize=11)   
ax2.set_ylabel("Y-axis", fontsize=11)
ax2.set_aspect('equal', ) #set equal aspect ratio for x and y axes

plt.tight_layout() #adjust subplots to fit into figure area.
plt.show()  
```
