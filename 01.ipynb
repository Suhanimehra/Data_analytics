```python
import matplotlib.pyplot as plt
import numpy as np

incomes = np.random.normal(loc=100.0 , scale=20.0 ,size=1000)
plt.hist(incomes,bins=50)

plt.show()
```

```python
print(np.mean(incomes))
print(np.median(incomes))
```

```python
new_incomes=np.random.normal(loc=27000,scale=15000,size=1000)
plt.hist(new_incomes,bins=50)
plt.show()
```

```python
print(np.mean(new_incomes))
print(np.median(new_incomes))
```

```python
new_incomes=np.append(new_incomes,[10000000000])
# this will give mean after adding extreme values.
print(np.mean(new_incomes))
print(np.median(new_incomes))
```

```python
ages=np.random.randint(low=18,high=90,size=500)
print(ages)
```

```python
from scipy import stats
from collections import Counter
print(stats.mode(ages))

age_count=Counter(ages)
print(age_count.most_common(1))
```
