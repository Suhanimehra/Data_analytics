```python
print("Negative beer servings rows " , df[df['beer_servings']<0][['country','beer_servings']])
beer_servings_negative_count=(df['beer_servings']<0).sum()
print("Negative beer servings count: ",beer_servings_negative_count)
```

### Lets Fix Negative, either feed 0 or filter out countries with negative servings

```python
df_copy=df.copy()
print("Df dimensions before removing -beer servings " , df.shape)
df = df[df['beer_servings']>=0]
print("Df dimensions after removing -beer servings " , df.shape)
```

```python
#  to feed 0 , use loc function and feed the value at the index and column name 
```

```python
# Get all feature names from the DataFrame
feature_name = df.columns
print(feature_name)

# Remove leading/trailing spaces and convert to title case
df['continent'] = df['continent'].str.strip().str.title()

print(df['continent'].head())
```

```python
print(df['continent'].unique())
```

# Grouping and Aggregation

```python
avg_beer_servings = df.groupby('continent')['beer_servings'].mean().round(2)
print(avg_beer_servings)
```
