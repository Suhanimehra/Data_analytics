Standard Deviation

```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
incomes=np.random.normal(100,20.0,10000)
plt.hist(incomes,bins=50)
plt.show()
```

Population Standard Deviation:

σ = √[ Σ(xᵢ - μ)² / N ]

 what is standard deviation? 
 Standard deviation is a measure of the amount of variation or dispersion in a set of values. 

 steps for std
 1. Calculate the mean of the data set.
 2. Subtract the mean from each data point and square the result.
 3. Calculate the average of these squared differences.
 4. Take the square root of this average to get the standard deviation.

```python
from scipy import stats

print(incomes.std())

#using stats
import statistics
print("using statistics",statistics.stdev(incomes))

#using scipy
print("using scipy",stats.tstd(incomes))

#using numpy
print(np.std(incomes))
```

Variance

what is variance?
 Variance is a measure of how much the values in a data set differ from the mean of the data set. 

steps for variance
 1. Calculate the mean of the data set.
 2. Subtract the mean from each data point and square the result.
 3. Calculate the average of these squared differences.

```python
from scipy import stats
import statistics

print(incomes.var()) #var is for varience

#using stats

print(statistics.variance(incomes))

#using scipy

print(stats.tvar(incomes))
```

```python
data=[1,4,5,4,8]

mean=sum(data)/len(data)
print(mean)

differences=[x-mean for x in data] #cqalculate the fifference for each number in list
print(differences)

sqaured_distances=[d**2 for d in differences]
print(sqaured_distances)

variance=sum(sqaured_distances)/len(data)
print(variance)


#stats

variance_stats=statistics.variance(data)
print(variance_stats)

#scipy

variance_scipy=stats.tvar(data)
print(variance_scipy)

#numpy

variance_numpy=np.var(data)
print(variance_numpy)
```
