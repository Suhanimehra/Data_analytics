### 1. Formulating Hypotheses from Visulations 



```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
from scipy import stats 
```

```python
np.random.seed(42)
df_hyp=pd.DataFrame({
    'device' : ['mobile'] *100 + ['desktop']*100,
    'conversion' : np.concatenate([np.random.binomial(1,0.05,100), np.random.binomial(1,0.12,100)])
})

print(df_hyp)
conversion_rate_by_device = df_hyp.groupby('device')['conversion'].mean()
print("Conversion rate by device : " ,conversion_rate_by_device)

# Visualization 
plt.figure(figsize=(10,4))
sns.barplot(x='device' , y = 'conversion' , hue='device',data=df_hyp, palette='Blues_d',legend = False , errorbar=None)
plt.title("Visual Hypothesis : Conversion Rate b Device Type")
plt.ylabel('Conversion Rate')
plt.show()
```

```python
np.random.seed(42)

control_group = np.random.normal(loc=125, scale = 8 , size = 25)
print("Control Group: " , control_group)

camp_group = np.random.normal(loc=135 , scale = 8 , size = 25)
print("Camp Group:", camp_group)

#Formulating Hypothesis
#Null Hypothesis(H0): 
#Average Hypothesis(H1):

#run indepedent one side t-test(testing if the treatment group has a higher average strike rate)


t_statistics,p_value = stats.ttest_ind(camp_group,control_group,alternative='greater')

print(f"Average Strike Rate: -Control Group: {np.mean(control_group):.2f}, camp_group: , {np.mean(camp_group):.2f}")
print("-"*80)

print(f"T-Stats: , {t_statistics:.2f} , P_value, {p_value:.2f}")


if p_value <0.05:
    print("Reject Null Hypothesis: The treatment group has a higher average strike rate than the control group.")
else:
    print("Fail to Reject Null Hypothesis: The treatment group does not have a higher average strike rate than the control group.")



 
```
